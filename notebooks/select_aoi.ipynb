This notebook adjust the AOI of my project to use a smaller amount of data

```python
import geopandas as gpd
from shapely.geometry import Polygon
```

```python

# Define your polygon vertices as (longitude, latitude) pairs
# Note: X = longitude, Y = latitude
polygon_coords = [
    (28.9151330, -1.9738515),
    (28.9006982, -1.9724208),
    (28.9019598, -1.9825989),
    (28.9075415, -1.9954450),
    (28.8883063, -2.0113312),
    (28.9031538, -2.0283872),
    (28.9089273, -2.0207053),
    (28.9150809, -1.9972770),
    (28.9188259, -1.9933761),
    (28.9132455, -1.9853107),
    (28.9151330, -1.9738515)
    # ... add as many points as you need ...
    # (Optional: the polygon will auto-close, but repeating the first point is safe)
]

# 1. Create the GeoDataFrame in geographic coordinates (EPSG:4326)
aoi_wgs84 = gpd.GeoDataFrame(geometry=[Polygon(polygon_coords)], crs="EPSG:4326")

# 2. Transform/reproject it directly into your CRS (EPSG:32735 - UTM zone 35S)
aoi = aoi_wgs84.to_crs("EPSG:32735")

print(aoi)
```
